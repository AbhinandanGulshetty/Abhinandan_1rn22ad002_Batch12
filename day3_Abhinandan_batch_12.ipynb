### `sleep_in`

```python
def sleep_in(weekday, vacation):
  return not weekday or vacation
```

### `monkey_trouble`

```python
def monkey_trouble(a_smile, b_smile):
  return a_smile == b_smile
```

### `sum_double`

```python
def sum_double(a, b):
  sum = a + b
  if a == b:
    sum = sum * 2
  return sum
```

### `diff21`

```python
def diff21(n):
  if n <= 21:
    return 21 - n
  else:
    return (n - 21) * 2
```

### `parrot_trouble`

```python
def parrot_trouble(talking, hour):
  return talking and (hour < 7 or hour > 20)
```

### `makes10`

```python
def makes10(a, b):
  return (a == 10 or b == 10 or a + b == 10)
```

### `near_hundred`

```python
def near_hundred(n):
  return ((abs(100 - n) <= 10) or (abs(200 - n) <= 10))
```

### `pos_neg`

```python
def pos_neg(a, b, negative):
  if negative:
    return (a < 0 and b < 0)
  else:
    return ((a < 0 and b > 0) or (a > 0 and b < 0))
```

### `not_string`

```python
def not_string(str):
  if len(str) >= 3 and str[:3] == "not":
    return str
  return "not " + str
```

### `missing_char`

```python
def missing_char(str, n):
  return str[:n] + str[n+1:]
```

### `front_back`

```python
def front_back(str):
  if len(str) <= 1:
    return str
  mid = str[1:len(str)-1]
  return str[len(str)-1] + mid + str[0]
```

### `front3`

```python
def front3(str):
  return str[:3] * 3
```
