### Q1) Check if a number is prime

```python
def is_prime(num):
    if num <= 1:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True

# Get input from the user
try:
    number = int(input("Enter an integer to check if it's prime: "))
    if is_prime(number):
        print(f"{number} is a prime number.")
    else:
        print(f"{number} is not a prime number.")
except ValueError:
    print("Invalid input. Please enter an integer.")
```

### Q2) Simple calculator - take input from user

```python
def calculator():
    try:
        num1 = float(input("Enter the first number: "))
        operator = input("Enter an operator (+, -, *, /): ")
        num2 = float(input("Enter the second number: "))

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 == 0:
                print("Error: Division by zero is not allowed.")
                return
            result = num1 / num2
        else:
            print("Error: Invalid operator.")
            return

        print(f"The result is: {result}")
    except ValueError:
        print("Invalid input. Please enter valid numbers.")

calculator()
```

### Q3) Check for the leap year

```python
def is_leap_year(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return True
    return False

# Get input from the user
try:
    year = int(input("Enter a year to check if it's a leap year: "))
    if is_leap_year(year):
        print(f"{year} is a leap year.")
    else:
        print(f"{year} is not a leap year.")
except ValueError:
    print("Invalid input. Please enter an integer for the year.")
```

### Q4) Check for username & password if they are correct login else error

```python
def login_system():
    # Predefined username and password
    correct_username = "user123"
    correct_password = "pass456"

    # Get input from the user
    username = input("Enter username: ")
    password = input("Enter password: ")

    if username == correct_username and password == correct_password:
        print("Login successful! Welcome.")
    else:
        print("Error: Incorrect username or password. Please try again.")

login_system()
```
